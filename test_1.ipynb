```python
import os 
from dotenv import load_dotenv 
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY" )
```

```python
import os 
from langchain.google_genai import ChatGoogleGenerativeAI

model= ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
response = model.invoke("why do parrots talk?")
response
```

```python
for chunk in model.stream("wirte me a 200 words paragraph on why parrots talk"):
    print(chunk, end="|", flush= True)
```

```python
responses = model.batch([
    "write me a 200 words paragraph on why parrots talk", 
    "write me a 200 words paragraph on why we brush daily",
    "write me a 200 words paragraph on why sky is blue"
])

for response in responses:
    print(response)
```
